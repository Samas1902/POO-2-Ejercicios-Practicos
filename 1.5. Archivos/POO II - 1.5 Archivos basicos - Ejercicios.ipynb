## 1. Abrir un archivo con `with open(...)`

Para trabajar con un archivo se usa `open(nombre, modo)`. Escribirlo dentro de un bloque `with` hace que Python **cierre el archivo automáticamente** al terminar, como el *try-with-resources* de Java.

| Modo | Significa | Si el archivo ya existe | Si no existe |
|---|---|---|---|
| `"w"` | escribir (*write*) | **borra** su contenido | lo crea |
| `"r"` | leer (*read*) | lo lee | error `FileNotFoundError` |
| `"a"` | agregar (*append*) | escribe al final | lo crea |

Siempre agregue `encoding="utf-8"` para que las tildes y la ñ se guarden bien.

## 2. Escribir en un archivo (modo `"w"`)
`write()` no salta de línea por sí solo: hay que poner `\n` al final de cada línea.

````python
with open("saludo.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Hola, estudiantes de POO II\n")
    archivo.write("Este archivo lo creó Python\n")
    archivo.write("Cuenca - Ecuador\n")
    archivo.write("Johncito es tetoncito\n")
    archivo.write("Dieguito cochinito\n")

print("Archivo creado")
````

````python

with open("asignaturas.txt", "w", encoding="utf-8") as archivo:
    archivo.write("POO II\n")
    archivo.write("Base de datos II\n")
    archivo.write("Desarrollo de APPs web\n")
    archivo.write("Expresion oral y escrita\n")
    archivo.write("Estadística\n")
    lista_asignaturas = ["POO II", "Base de datos II", "Desarrollo de APPs web", "Estadística", "Expresion Oral y Escrita"]
    for a in lista_asignaturas:
        archivo.write(a + "\n")

print("Se creó el archivo")
````

````python
with open("persona_diccionario.txt", "w", encoding="utf-8") as archivo:
    dic_persona = {
        "Nombre" : "Samuel",
        "Apellido" : "Tobar",
        "Edad" : 18,
        "Correo" : "ola123@gmail.com"
    }
    for key,valor in dic_persona.items():
        archivo.write(f"{key}: {valor}\n")

print("Exito")
````

````python
cinco_personas = []
persona1 = {
    "Nombre" : "Samuel",
    "Apellido" : "Tobar",
    "Edad" : 18,
    "Correo" : "ola123@gmail.com"
}
persona2 = {
    "Nombre" : "Johncito",
    "Apellido" : "Cabrera",
    "Edad" : 24,
    "Correo" : "ola123@gmail.com"
}
persona3 = {
    "Nombre" : "Santiago",
    "Apellido" : "Rubio",
    "Edad" : 24,
    "Correo" : "ola123@gmail.com"
}
persona4 = {
    "Nombre" : "Diego",
    "Apellido" : "Rojas",
    "Edad" : 19,
    "Correo" : "ola123@gmail.com"
}
persona5 = {
    "Nombre" : "Deivid",
    "Apellido" : "Ortega",
    "Edad" : 19,
    "Correo" : "ola123@gmail.com"
}
cinco_personas.append(persona1)
cinco_personas.append(persona2)
cinco_personas.append(persona3)
cinco_personas.append(persona4)
cinco_personas.append(persona5)

def bonito(diccionario):
    lineas = []
    for key, valor in diccionario.items():
        lineas.append(f"    {key}: {valor}")

    return "\n".join(lineas)

with open ("cinco_personas.txt", "w", encoding="utf-8") as archivo:
    i = 1
    for persona in cinco_personas:
        archivo.write(f"Persona {i}: \n")
        archivo.write(f"{bonito(persona)} \n")
        i+=1

print("Exito")
````

## 3. Leer un archivo completo (modo `"r"`)
`read()` devuelve todo el contenido en un solo texto.

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
````

## 4. Leer línea por línea
Un archivo se puede recorrer con `for`, igual que una lista. Cada línea trae su `\n` al final; `strip()` lo quita.

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    numero = 1
    for linea in archivo:
        print(numero, "→", linea.strip())
        numero += 1
````

````python
with open("cinco_personas.txt", "r", encoding="utf-8") as archivo:
    numero = 1
    for linea in archivo:
        print(numero, "→", linea.strip())
        numero += 1
````

También se pueden obtener todas las líneas en una lista con `readlines()`:

````python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

print(lineas)
print("El archivo tiene", len(lineas), "líneas")
````

````python
with open("cinco_personas.txt", "r", encoding="utf-8") as archivo:
    lineas = archivo.readlines()

print(lineas)
print("El archivo tiene", len(lineas), "líneas")
````

## 5. Agregar sin borrar (modo `"a"`)
Con `"w"` se pierde lo anterior; con `"a"` lo nuevo se añade al final.

````python
with open("saludo.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Línea agregada después\n")

with open("saludo.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````

````python
with open("cinco_personas.txt", "a", encoding="utf-8") as archivo:
    archivo.write("Persona 6:\n")
    archivo.write("    Nombre: Gerard\n")
    archivo.write("    Apellido: Pérez\n")
    archivo.write("    Edad : 20\n")
    archivo.write("    Correo: ola123@gmail.com\n")
````

## 6. Guardar datos separados por comas
Cada línea es un registro y cada dato se separa con una coma. Para escribir se unen los datos con `f-string`; para leer se separan con `split(",")`.

````python
estudiantes = [("Ana", 9.5), ("Luis", 7.0), ("María", 8.25), ("Samu", 10), ("Johncito",0)]

with open("notas.txt", "w", encoding="utf-8") as archivo:
    for nombre, nota in estudiantes:
        archivo.write(f"{nombre},{nota}\n")
suma = 0
i = 0
with open("notas.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        datos = linea.strip().split(",")
        nombre = datos[0]
        suma += nota
        i += 1
        nota = float(datos[1])      # lo que se lee de un archivo siempre es texto
        print(f"{nombre:<8} {nota:>5.2f}")

promedio = suma/i
print(f"El promedio es: {promedio:.2f}")
````

> **Importante:** todo lo que se lee de un archivo es texto (`str`). Para hacer cálculos hay que convertirlo con `int()` o `float()`.

## 7. ¿Y si el archivo no existe?
Leer un archivo que no existe produce `FileNotFoundError` y el programa se detiene. Con `try / except` se muestra un mensaje y el programa continúa.

````python
try:
    with open("no_existe.txt", "r", encoding="utf-8") as archivo:
        print(archivo.read())
except FileNotFoundError:
    print("El archivo no existe, revise el nombre")

print("El programa sigue funcionando")
````

## Resumen

| Quiero… | Código |
|---|---|
| Crear o reemplazar | `with open("a.txt", "w", encoding="utf-8") as f:` `f.write("texto\n")` |
| Leer todo | `f.read()` |
| Leer línea por línea | `for linea in f:` `linea.strip()` |
| Agregar al final | `open("a.txt", "a", encoding="utf-8")` |
| Separar datos | `linea.strip().split(",")` |
| Archivo que no existe | `try:` … `except FileNotFoundError:` |

### Ejercicio 8. Promedio del curso
Lea `calificaciones.txt` y calcule el **promedio** de las notas. Muestre también cuántos estudiantes **aprobaron** (nota mayor o igual a 7).

````python
# Escriba su código aquí
with open("calificaciones.txt", "r", encoding="utf-8") as archivo:
    i = 0
    suma = 0
    aprobaron = 0
    for linea in archivo:
        suma += int(linea.split(",")[1])
        if int(linea.split(",")[1]) >= 7:
            aprobaron += 1
        i += 1
        
print(f"El número de estudiantes que aprobaron es: {aprobaron}")
print(f"El promedio de las calificaciones es: {suma/i}")
````

````python
# Escriba su código aquí
with open("calificaciones.txt", "r", encoding="utf-8") as archivo1, open("aprobados.txt", "w", encoding="utf-8") as archivo2:
    for linea in archivo1:
        if int(linea.split(",")[1]) >= 7:
            archivo2.write(linea)

with open("aprobados.txt", "r", encoding="utf-8") as archivo:
    print(archivo.read())
````
